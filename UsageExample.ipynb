## 0. 先认识一下这个页面

Jupyter Notebook 由一个个**单元格（Cell）**从上到下堆叠而成，主要有两种类型：

| 类型 | 作用 | 外观标记 |
|------|------|----------|
| Markdown | 写文字、标题、表格 | 左侧**没有** `In [ ]` 编号 |
| Code | 写并运行 Python 代码 | 左侧**有** `In [ ]` 编号 |

**如何运行一个单元格？**
- 点击选中单元格，按 **`Shift + Enter`** → 运行并跳到下一个单元格
- 按 **`Ctrl + Enter`** → 只运行当前单元格，不跳转

运行后，左侧的 `In [ ]` 会变成 `In [1]`、`In [2]`……编号表示执行的先后顺序。

## 1. 经典 Hello, World! 👋

选中下面这个单元格，按 `Shift + Enter` 运行试试：

```python
print("Hello, World!")
```

上面代码输出了 `Hello, World!`——这是 Python 最经典的入门程序。

现在我们加点花样，用变量和循环来问候：

```python
name = "Jupyter"
print(f"Hello, {name}!")
print("=" * 30)
for i in range(1, 6):
    print("Hello, World! 第", i, "次")
```

> 💡 **关键点**：变量 `name` 定义在上一个单元格里，但下面的单元格依然能使用它。
> 因为**所有 Code 单元格共享同一个 Python 内核**。这也正是你前面那个问题——「把 `.py` 代码分段搬进 Notebook 还能跑吗」的答案：能跑，只要按顺序执行、且不把一个函数定义拆到两个单元格。

```python
# 验证：上一个单元格定义的 name 在这里还能用
print("name 变量的值还是：", name)
```

## 2. 在 Notebook 里画图 📈

Notebook 的一大优势就是**图文并茂**：用 matplotlib 画的图会直接显示在单元格下方，无需弹出新窗口。

```python
import matplotlib.pyplot as plt
import numpy as np

# 生成 0 到 2π 之间的 100 个点
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label="sin(x)", color="tab:blue", linewidth=2)
plt.title("Hello, Plot! — 一个正弦曲线")
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 3. 魔法命令（Magic Commands）✨

Jupyter 提供一些以 `%` 开头的「魔法命令」，用来查看环境、计时等，非常方便：

```python
# %time 计算代码运行耗时
%time sum(range(10**7))
```

```python
# %who 查看当前已经定义了哪些变量
%who
```

```python
# %ls 查看当前目录下的文件（相当于 shell 的 ls）
%ls
```

## 4. 常用快捷键速查表

| 操作 | 快捷键 |
|------|--------|
| 运行当前单元格并跳到下一个 | `Shift + Enter` |
| 运行当前单元格不跳转 | `Ctrl + Enter` |
| 在下方插入新单元格 | `B` |
| 在上方插入新单元格 | `A` |
| 删除当前单元格 | 双击右侧 → `D D` |
| 切换为 Code 单元格 | `Y` |
| 切换为 Markdown 单元格 | `M` |
| 重启并运行全部 | 菜单 `Kernel → Restart & Run All` |

> 提示：按 `Esc` 进入**命令模式**（单元格左边框变蓝）后，才能用 `A/B/M/Y/D` 这些字母快捷键；
> 按 `Enter` 进入**编辑模式**（边框变绿）开始打字。
